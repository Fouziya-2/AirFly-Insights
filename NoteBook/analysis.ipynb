```python
# selecting relevent columns for analysis after loadings the df selecting required cols
selected_cols=["FL_DATE","OP_CARRIER","ORIGIN","DEST","CRS_DEP_TIME","DEP_TIME","DEP_DELAY",
               "ARR_TIME","ARR_DELAY","CANCELLED","CANCELLATION_CODE","DISTANCE","CRS_DEP_TIME","CRS_ARR_TIME","TAXI_IN","TAXI_OUT"]
df=df[selected_cols]
df.shape
```

```python
# inspecting the data structure using
df.info()
```

```python
# formating Date column
df['FL_DATE ']=pd.to_datetime(df['FL_DATE'])
```

```python
# checking for null values and sum of null
df.isnull().sum()
```

```python
# checking % of null values analysis
(df.isnull().sum()/len(df))*100
```

```python
#  droping the high % of null values columns
df=df.drop(columns=['CANCELLATION_CODE'])
```

```python
# columns
df.columns
```

```python
# checking the % analysis again after removed columns
(df.isnull().sum()/len(df))*100
```

```python
# filtering the cancelled flights 0=flights was not cancel,1=flights cancelled
df["CANCELLED"].value_counts()
```

```python
# after filtering using independent df for removing cancelled flights
df=df[df['CANCELLED']==0].copy()
```

```python
df.shape
```

```python
#  again checking %
(df.isnull().sum()/len(df))*100
```

```python
# memory optimization
df.memory_usage(deep=True).sum()/1024**2
```

```python
# converting date column properly
df['FL_DATE']=pd.to_datetime(df['FL_DATE'])
```

```python
# convering category cols
df['OP_CARRIER'] = df['OP_CARRIER'].astype('category')
df['ORIGIN'] = df['ORIGIN'].astype('category')
df['DEST'] = df['DEST'].astype('category')
```

```python
# reduce numeric precision  instead of float64_ used 32
float_cols = ['DEP_DELAY', 'ARR_DELAY', 'TAXI_OUT', 'TAXI_IN']

for col in float_cols:
    df[col] = df[col].astype('float32')
```

```python
df.memory_usage(deep=True).sum()/1024**2
```

```python
# understanding the patterns and statistical distribution
df.describe()
```

```python
# I used median instead of mean because flight delay data is 
# skewed and contains extreme outliers. Median gives 
# a more robust measure of central tendency.
df['DEP_DELAY'].median()
df['ARR_DELAY'].median()
```

```python
df.duplicated().sum()
```

```python
# data sampling for 20% data
df_sample=df.sample(frac=0.2,random_state=42)
df_sample.shape
```

```python
df.dtypes
```

```python
df.memory_usage(deep=True).sum()/1024**2
```

```python
df['CANCELLED']=df['CANCELLED'].astype('Int8')
```

```python
df.dtypes
```

```python
# week-2 Analysis and feature analisis 
# flights are cancelled usually having missing delay/taxitimes taking consistant data before feature engineering
# Fill delay columns with median
delay_cols = ['DEP_DELAY', 'ARR_DELAY', 'TAXI_IN', 'TAXI_OUT']

for col in delay_cols:
    df.loc[:, col] = df[col].fillna(df[col].median())

# For cancelled flights, ensure delay = 0 if cancelled
df.loc[df['CANCELLED'] == 1, ['DEP_DELAY', 'ARR_DELAY']] = 0
```

```python
(df.isnull().sum() / len(df)) * 100
```

```python
df['FL_DATE']=pd.to_datetime(df['FL_DATE'])
```

```python
df.dtypes
```

```python
# month deriving features
df['MONTH']=df['FL_DATE'].dt.month
```

```python
# day of week(number) like 0=mon,6=sun
df['DAY_OF_WEEK'] = df['FL_DATE'].dt.dayofweek
```

```python
# day name(better for visualization)
df['DAY_NAME'] = df['FL_DATE'].dt.day_name()
```

```python
df.columns #duplicate cols have in that df
# df['DEP_HOUR'] = (df['CRS_DEP_TIME'] // 100).astype(int)
```

```python
# removing the duplicate columns
df = df.loc[:, ~df.columns.duplicated()]
```

```python
df.columns
```

```python
# departure hour,CRS_DEP_TIME hhmm format extracting hh hours
df['DEP_HOUR'] = (df['CRS_DEP_TIME'] // 100).astype(int)
```

```python
# extracting routes like below columns 
df['ROUTE'] = df['ORIGIN'].astype(str) + '-' + df['DEST'].astype(str)
```

```python
df[['ORIGIN', 'DEST', 'ROUTE']].head()
```

```python
# checking feature engineering columns
df[['FL_DATE','MONTH','DAY_OF_WEEK','DAY_NAME','DEP_HOUR','ROUTE']].head()
```

```python
# converting datatypes
df['MONTH'] = df['MONTH'].astype('int8')
df['DAY_OF_WEEK'] = df['DAY_OF_WEEK'].astype('int8')
df['DEP_HOUR'] = df['DEP_HOUR'].astype('int8')
df['DAY_NAME'] = df['DAY_NAME'].astype('category')
df['ROUTE'] = df['ROUTE'].astype('category')
```

```python
df.memory_usage(deep=True).sum()/1024**2
```
